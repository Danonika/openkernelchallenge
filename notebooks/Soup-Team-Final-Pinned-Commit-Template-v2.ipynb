```python
import os,sys,json,hashlib,subprocess,tempfile,re,base64,shutil
from pathlib import Path
from urllib.parse import urlsplit
assert isinstance(COMMIT,str) and re.fullmatch(r'[0-9a-fA-F]{40}',COMMIT), 'Set the real final Git commit first.'
url=urlsplit(REPO_URL)
assert url.scheme=='https' and url.netloc and url.username is None and url.password is None, 'Set the real HTTPS team repository URL without credentials.'
assert not Path(PROJECT_SUBDIRECTORY).is_absolute()
os.environ['CUDA_VISIBLE_DEVICES']='0'
os.environ['HF_HOME']='/kaggle/temp/soup-r026-hf'
os.environ['PYTHONUNBUFFERED']='1'
try:
    from kaggle_secrets import UserSecretsClient
    os.environ['HF_TOKEN']=UserSecretsClient().get_secret('HF_TOKEN')
except Exception:
    pass
workspace=Path(tempfile.mkdtemp(prefix='soup-team-final-git-',dir='/kaggle/working'))
root=workspace
def command(args,log,env=None):
    print('COMMAND',args,flush=True)
    with (root/log).open('w',encoding='utf-8') as out:
        process=subprocess.Popen(args,cwd=root,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,
                                 text=True,env=dict(os.environ,**(env or {})))
        for line in process.stdout:
            print(line,end='',flush=True);out.write(line);out.flush()
        result=process.wait()
    print('EXIT',result,log,flush=True)
    return result
assert command([sys.executable,'-m','pip','install','-q','soup-cli[train]==0.77.0',
    'transformers==5.16.1','peft==0.20.0','trl==0.29.1','bitsandbytes==0.50.2',
    'accelerate==1.14.0','pytest'],'setup.log')==0
token=''
try:
    from kaggle_secrets import UserSecretsClient
    token=UserSecretsClient().get_secret('GITHUB_TOKEN')
except Exception:
    pass
auth=[]
if token:
    encoded=base64.b64encode(('x-access-token:'+token).encode()).decode()
    auth=['-c','http.extraHeader=Authorization: Basic '+encoded]
checkout=workspace/'checkout'
clone=subprocess.run(['git',*auth,'clone','-q',REPO_URL,str(checkout)],capture_output=True,text=True)
del token,auth
if 'encoded' in globals():del encoded
assert clone.returncode==0, 'Read-only clone failed; check the URL and secret permissions.'
subprocess.run(['git','-C',str(checkout),'checkout','-q',COMMIT],check=True)
actual_commit=subprocess.run(['git','-C',str(checkout),'rev-parse','HEAD'],capture_output=True,text=True,check=True).stdout.strip()
assert actual_commit.lower()==COMMIT.lower()
root=(checkout/PROJECT_SUBDIRECTORY).resolve()
assert root.is_relative_to(checkout.resolve()) and (root/'solution').is_dir()
shutil.copyfile(workspace/'setup.log',root/'setup.log')
print('VERIFIED_FINAL_GIT_COMMIT',actual_commit,flush=True)
```
